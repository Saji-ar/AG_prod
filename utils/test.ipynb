```python
import subprocess
import os

def xlsx_to_pdf_with_libreoffice(xlsx_path: str):
    outdir = os.path.dirname(xlsx_path) or "."
    subprocess.run([
        "libreoffice",
        "--headless",
        "--convert-to", "pdf:writer_pdf_Export",
        "--outdir", outdir,
        xlsx_path
    ], check=True)
    return os.path.splitext(xlsx_path)[0] + ".pdf"


# … inside invoice_maker, after wb.save:
pdf_path = xlsx_to_pdf_with_libreoffice("data/test.xlsx")

print("PDF created at", pdf_path)
```

```python
import convertapi
import streamlit as st

convertapi.configure(api_key=st.secrets["convertapi_key"])

def convert_excel_to_pdf(excel_file):
    # Save uploaded file temporarily
    with open("temp.xlsx", "wb") as f:
        f.write(excel_file.getvalue())
    
    # Convert with formatting options
    result = convertapi.convert('pdf', {
        'File': 'temp.xlsx',
        'AutoFit': True,
        'PageOrientation': 'Portrait'
    }, from_format='xlsx')
    
    return result.file.io
r = convert_excel_to_pdf("data/test.xlsx")
```
